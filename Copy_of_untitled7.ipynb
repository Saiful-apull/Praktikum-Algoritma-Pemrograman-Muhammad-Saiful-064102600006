```python
while True:
  print("\n--- Program Penentu Jumlah Hari ---")
  try:
    bulan = int(input("Masukkan angka bulan (1-12) atau ketik 0 untuk keluar: "))

    # Kondisi untuk keluar dari perulangan
    if bulan == 0:
      print("Terima kasih telah menggunakan program ini!")
      break

    # Validasi input bulan
    if bulan < 1 or bulan > 12:
      print("Bulan tidak valid! Silakan masukkan angka antara 1 sampai 12.")
      continue

    tahun = int(input("Masukkan tahun: "))

    # Menentukan tahun kabisat
    is_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

    # Menentukan jumlah hari berdasarkan bulan
    if bulan in [1, 3, 5, 7, 8, 10, 12]:
      jumlah_hari = 31
    elif bulan in [4, 6, 9, 11]:
      jumlah_hari = 30
    elif bulan == 2:
      jumlah_hari = 29 if is_kabisat else 28

    print(
        f"Jumlah hari pada bulan {bulan} tahun {tahun} adalah {jumlah_hari}"
        " hari."
    )

  except ValueError:
    print("Input tidak valid! Harap masukkan angka yang berupa bilangan bulat.")
```
